## 1. Import Libraries
Pandas is used for data handling, NumPy for numerical operations, and Matplotlib/Seaborn for visualizations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
sns.set_theme(style="whitegrid")
```

## 2. Load Dataset

```python
file_path = "supermarket_sales_500_rows.csv"
df = pd.read_csv(file_path)

print("Dataset loaded successfully.")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
display(df.head())
```

## 3. Dataset Overview

```python
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nData types:")
print(df.dtypes)
```

## 4. Data Quality Check

```python
print("Missing values:")
display(df.isnull().sum())

print("Total missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

## 5. Data Cleaning

```python
df["Date"] = pd.to_datetime(df["Date"])
df = df.drop_duplicates().copy()

print("Shape after duplicate removal:", df.shape)
print("Missing values after cleaning:", df.isnull().sum().sum())
```

## 6. Verify Sales Calculation

```python
df["Calculated Sales"] = df["Quantity"] * df["Unit Price"]
matches = np.isclose(df["Sales"], df["Calculated Sales"], atol=0.01)

print("Matching Sales calculations:", matches.sum(), "out of", len(df))
display(df.loc[~matches, ["Invoice ID","Quantity","Unit Price","Sales","Calculated Sales"]].head(10))
```

## 7. Overall Sales Summary

```python
summary = pd.DataFrame({
    "Metric": [
        "Number of Transactions",
        "Total Quantity Sold",
        "Total Sales",
        "Average Transaction Sales",
        "Average Customer Rating"
    ],
    "Value": [
        len(df),
        df["Quantity"].sum(),
        df["Sales"].sum(),
        df["Sales"].mean(),
        df["Rating"].mean()
    ]
})
display(summary)
```

## 8. Product Analysis

```python
product_sales = df.groupby("Product")["Sales"].sum().sort_values(ascending=False)
display(product_sales.to_frame("Total Sales"))

print("Highest-sales product:", product_sales.idxmax())
print(f"Sales: ₹{product_sales.max():,.2f}")
```

```python
plt.figure(figsize=(10,6))
product_sales.sort_values().plot(kind="barh")
plt.title("Sales by Product")
plt.xlabel("Sales")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

## 9. Branch Analysis

```python
branch_sales = df.groupby(["Branch","City"])["Sales"].sum().sort_values(ascending=False)
display(branch_sales.to_frame("Total Sales"))

b = branch_sales.idxmax()
print(f"Best-performing branch: Branch {b[0]} ({b[1]})")
print(f"Sales: ₹{branch_sales.max():,.2f}")
```

```python
branch_chart = df.groupby(["Branch","City"])["Sales"].sum().reset_index()
branch_chart["Branch_City"] = branch_chart["Branch"] + " (" + branch_chart["City"] + ")"

plt.figure(figsize=(9,5))
sns.barplot(data=branch_chart, x="Branch_City", y="Sales")
plt.title("Sales by Branch")
plt.xlabel("Branch")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

## 10. Category Analysis

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
display(category_sales.to_frame("Total Sales"))
print("Highest-sales category:", category_sales.idxmax())
print(f"Sales: ₹{category_sales.max():,.2f}")
```

```python
plt.figure(figsize=(9,5))
sns.barplot(x=category_sales.values, y=category_sales.index)
plt.title("Sales by Category")
plt.xlabel("Sales")
plt.ylabel("Category")
plt.tight_layout()
plt.show()
```

## 11. Payment Method Analysis

```python
payment_counts = df["Payment"].value_counts()
display(payment_counts.to_frame("Transaction Count"))
print("Most-used payment method:", payment_counts.idxmax())
print("Transactions:", payment_counts.max())
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x=payment_counts.index, y=payment_counts.values)
plt.title("Payment Method Usage")
plt.xlabel("Payment Method")
plt.ylabel("Transactions")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

## 12. Member vs Normal Customers

```python
customer_avg = df.groupby("Customer Type")["Sales"].mean()
display(customer_avg.to_frame("Average Transaction Sales"))

print(f"Average Member transaction: ₹{customer_avg['Member']:,.2f}")
print(f"Average Normal transaction: ₹{customer_avg['Normal']:,.2f}")
```

```python
plt.figure(figsize=(7,5))
sns.barplot(x=customer_avg.index, y=customer_avg.values)
plt.title("Average Transaction Sales by Customer Type")
plt.xlabel("Customer Type")
plt.ylabel("Average Sales")
plt.tight_layout()
plt.show()
```

## 13. Customer Rating Analysis

```python
print(f"Average customer rating: {df['Rating'].mean():.2f} out of 5")
display(df["Rating"].describe())
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Rating"], bins=10, kde=True)
plt.title("Distribution of Customer Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Transactions")
plt.tight_layout()
plt.show()
```

## 14. Additional Analysis

```python
print("Sales by City:")
display(df.groupby("City")["Sales"].sum().sort_values(ascending=False).to_frame("Total Sales"))

print("Sales by Gender:")
display(df.groupby("Gender")["Sales"].sum().sort_values(ascending=False).to_frame("Total Sales"))

print("Top 10 Transactions:")
display(df[["Invoice ID","Product","Branch","City","Quantity","Sales"]]
        .sort_values("Sales", ascending=False).head(10))
```

## 15. Business Insights

```python
print(f"1. {product_sales.idxmax()} generates the highest product sales: ₹{product_sales.max():,.2f}.")
b = branch_sales.idxmax()
print(f"2. Branch {b[0]} ({b[1]}) has the highest branch sales: ₹{branch_sales.max():,.2f}.")
print(f"3. {category_sales.idxmax()} is the highest-sales category: ₹{category_sales.max():,.2f}.")
print(f"4. {payment_counts.idxmax()} is the most-used payment method with {payment_counts.max()} transactions.")
print(f"5. Member average transaction: ₹{customer_avg['Member']:,.2f}; Normal average: ₹{customer_avg['Normal']:,.2f}.")
print(f"6. Average customer rating: {df['Rating'].mean():.2f}/5.")
```
